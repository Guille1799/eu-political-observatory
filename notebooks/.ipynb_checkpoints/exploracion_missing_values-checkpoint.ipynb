```python
import pandas as pd

# Cargar los tres archivos de ARDECO
df_unemployment = pd.read_csv('../data/raw/ARDECO-RPUCNP.versions_2024-unit_PC-level_id_2.table.csv')
df_gdp = pd.read_csv('../data/raw/_ARDECO-SUVGDP.versions_2024-unit_EUR,PPS_EU27_2020-level_id_0,1,2,3.table.csv_')
df_education = pd.read_csv('../data/raw/_ARDECO-RPDTN.versions_2024-unit_PC-isced11_ED0-2,ED3_4,ED5-8-age_Y25-64-sex_TOTAL-level_id_0,1,2.table.csv_')

print("Archivos cargados correctamente")
```

```python
# Analizar valores vacíos en cada dataset
print("=== DESEMPLEO ===")
print(df_unemployment.isnull().sum())
print(f"Total filas: {len(df_unemployment)}")

print("\n=== PIB ===")
print(df_gdp.isnull().sum())
print(f"Total filas: {len(df_gdp)}")

print("\n=== EDUCACIÓN ===")
print(df_education.isnull().sum())
print(f"Total filas: {len(df_education)}")
```

```python
# Ver qué regiones tienen huecos en 2008
print("=== REGIONES CON HUECOS EN DESEMPLEO 2008 ===")
print(df_unemployment[df_unemployment['2008'].isnull()][['TERRITORY_ID', 'NAME_HTML']])

print("\n=== REGIONES CON HUECOS EN PIB 2008 ===")
df_gdp_nuts2 = df_gdp[df_gdp['LEVEL_ID'] == 2]
print(df_gdp_nuts2[df_gdp_nuts2['2008'].isnull()][['TERRITORY_ID', 'NAME_HTML']])

print("\n=== REGIONES CON HUECOS EN EDUCACIÓN 2008 ===")
df_education_nuts2 = df_education[df_education['LEVEL_ID'] == 2]
print(df_education_nuts2[df_education_nuts2['2008'].isnull()][['TERRITORY_ID', 'NAME_HTML']])
```

```python
COUNTRIES = ['ES', 'DE', 'FR', 'IT', 'PL', 'HU', 'SE']

print("=== HUECOS EN DESEMPLEO 2008 — SOLO NUESTROS 7 PAÍSES ===")
df_unemp_filtered = df_unemployment[df_unemployment['TERRITORY_ID'].str[:2].isin(COUNTRIES)]
print(df_unemp_filtered[df_unemp_filtered['2008'].isnull()][['TERRITORY_ID', 'NAME_HTML']])

print("\n=== HUECOS EN PIB 2008 — SOLO NUESTROS 7 PAÍSES ===")
df_gdp_filtered = df_gdp[(df_gdp['LEVEL_ID'] == 2) & (df_gdp['TERRITORY_ID'].str[:2].isin(COUNTRIES))]
print(df_gdp_filtered[df_gdp_filtered['2008'].isnull()][['TERRITORY_ID', 'NAME_HTML']])

print("\n=== HUECOS EN EDUCACIÓN 2008 — SOLO NUESTROS 7 PAÍSES ===")
df_edu_filtered = df_education[(df_education['LEVEL_ID'] == 2) & (df_education['TERRITORY_ID'].str[:2].isin(COUNTRIES))]
print(df_edu_filtered[df_edu_filtered['2008'].isnull()][['TERRITORY_ID', 'NAME_HTML']])
```
